```python
import numpy as np
```

```python
v1 = np.array([2/3, 1/3, -2/3])
v2 = np.array([1/3, 2/3, 2/3])
v3 = np.array([2/3, -2/3, 1/3])
B = [v1, v2, v3]
```

```python
#check if the norm of each vector is 1
unitVectors = True
for vector in B:
    if np.linalg.norm(vector) != 1.:
        unitVectors = False
print(f"Are unit? {unitVectors}")
```

```python
#all pairwise orthogonal
dim = range(len(B))
orthogonalVectors = True
for i in dim:
    for j in dim:
        if (i != j) and np.dot(B[i], B[j]) != 0:
            orthogonalVectors = False

print(f'Are orthogonal? {orthogonalVectors}')
```
