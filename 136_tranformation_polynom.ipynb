F(p(t)) = [p(0), p(0) + p(1), p'(-1)] = [1, 2, 7]. Find the polynomial.

```python
import numpy as np
```

```python
F = np.array([
    [1, 0, 0],
    [2, 1, 1],
    [0, 1, -2]
])
F
```

```python
F_inverse = np.linalg.inv(F)
F_inverse
```

```python
polynom = np.matmul(F_inverse, [1, 7, 2])
polynom
```

Hence, p(t) = 1 + 4t + 1t^2
