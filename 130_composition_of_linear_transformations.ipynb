A reflection about the line y = x, followed by an orthogonal projection on the y-axis, followed by 60 degrees rotation (anticlockwise).

```python
#Reflectin about the the line y = x 

Sy = np.array([
    [0, 1],
    [1, 0]
])
Sy
```

```python
#Orthogonal projection on y-axis
Py = np.array([
    [0, 0],
    [0, 1]
])
Py
```

```python
#Rotation by 60 degrees
alpha = np.pi / 3
R = np.array([
    [np.cos(alpha), -np.sin(alpha)],
    [np.sin(alpha), np.cos(alpha)]
])
R
```

```python
H = np.matmul(R, np.matmul(Py, Sy))
H
```
