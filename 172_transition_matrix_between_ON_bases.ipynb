```python
import numpy as np
```

```python
round = np.round
array = np.array
norm = np.linalg.norm
sqrt = np.sqrt
```

```python
# First orthonormal basis B1
e1 = array([2/3,  2/3, 1/3])
e2 = array([-2/3, 1/3, 2/3])
e3 = array([1/3, -2/3, 2/3])
B1 = np.column_stack((e1, e2, e3))
B1
```

```python
# Second orthonormal basis (rotated standard basis)
f1 = array([1/sqrt(2),   1/sqrt(2),  0.0])
f2 = array([-1/sqrt(6),  1/sqrt(6),  2/sqrt(6)])
f3 = array([1/sqrt(3),  -1/sqrt(3),  1/sqrt(3)])
B2 = np.column_stack((f1, f2, f3))
B2
```

```python
# Verify B1 and B2 are orthonormal
# Check if B2^T @ B2 equals the Identity Matrix
if np.allclose(B1.T @ B1, np.eye(B1.shape[1]),atol=1e-4):
    print("Basis B1 is orthonormal")
if np.allclose(B2.T @ B2, np.eye(B2.shape[1]),atol=1e-4):
    print("Basis B2 is orthonormal")
```

```python
# Transition matrix from B2 -> B1
T21= B1.T @ B2
print("Transition Matrix T21 (B2 -> B1):")
print(np.round(T21, 3))
```

```python
# Check that matrix from B2 -> B1 is orhonormal

if np.allclose(T21.T @ T21, np.eye(T21.shape[1]),atol=1e-4):
    print("Basis T21 is orthonormal")
```

```python
#Transition matrix from B1 -> B2
T12= B2.T @ B1
print("Transition Matrix T12 (B1 -> B2):")
print(np.round(T12, 3))
```

```python
# Check that matrix from B1 -> B2 is orhonormal

if np.allclose(T12.T @ T12, np.eye(T12.shape[1]),atol=1e-4):
    print("Basis T12 is orthonormal")
```

```python
# Transform Vector Coordinates
v_in_B2 = np.array([2.0, -3.0, 1.0]) # Define a vector in B2 coordinates
v_in_B1 = T21 @ v_in_B2 # Convert to B1 coordinates
v_back_in_B2 = T21.T @ v_in_B1 # Convert back to B2 coordinates
print(f"\nCoordinates in B2:           {v_in_B2}")
print(f"Coordinates in B1:           {np.round(v_in_B1, 3)}")
print(f"Recovered coordinates in B2: {np.round(v_back_in_B2, 3)}")
```
