```python
import numpy as np
```

```python
A = np.array(
    [
        [2/3, 1/3, 2/3],
        [1/3, 2/3, -2/3],
        [-2/3, 2/3, 1/3]
    ]
)
A
```

```python
Atr = np.transpose(A)
Atr
```

```python
if np.array_equal(A @ Atr, np.identity(3)):
    print('Matrix is orthogonal')
```

```python
np.array_equal(Atr @ A, A @ Atr)
```
