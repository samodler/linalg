F: R3->R3
defined geometrically as a projection onto the plane x + y + 2z = 0
along the line (t, -t, t), where t is in R.

Determine the matrix of F in the standard basis of R3

```python
import numpy as np
import array_to_latex as a2l
```

Consider basis B = {v1, v2, v3}
where v1 and v2 belong to the plane, and v3 is the directional vector of the line (t, -t, t)

```python
#coordinates of the vectors forming basis B with respect to the standard basis S

v1_s = np.array(
    [1, -1, 0]
)
v2_s = np.array(
    [-2, 0, 1]
)
v3_s = np.array(
    [1, -1, 1]
)
```

```python
#transition matrix B->S
transition_b_to_s = np.array((v1_s, v2_s, v3_s)).transpose()
transition_b_to_s
```

```python
#transition matrix S->B
transition_s_to_b = np.linalg.inv(transition_b_to_s)
transition_s_to_b
```

```python
#matrix M representing the projection of a vector in R3 onto the plane with regard to the basis B, i.e. M is matrix of F in B.
#Since v1 and v2 belong to the plane their projection is equal to the original vector.
#Since v3 passes through the origin, its projection is the zero vector.
v1_projected = np.array([1, 0, 0])
v2_projected = np.array([0, 1, 0])
v3_projected = np.array([0, 0, 0])

M = np.array([v1_projected, v2_projected, v3_projected]).transpose()
M
```

```python
#F is matrix of F in S
F = transition_b_to_s @ M @ transition_s_to_b
F
```
