```python
# matrix for transformation that projects entire R3 on vector u
```

```python
u = np.array([1,2,3])
```

```python
e1 = np.array([1,0,0])
e2 = np.array([0,1,0])
e3 = np.array([0,0,1])
```

```python
proj_u_e1 = np.dot(e1,u) / np.dot(u,u) * u
proj_u_e2 = np.dot(e2,u) / np.dot(u,u) * u
proj_u_e3 = np.dot(e3,u) / np.dot(u,u) * u
```

```python
T = np.array([proj_u_e1, proj_u_e2, proj_u_e3]).transpose()
T
```

```python
# verify that vector v1-3 form ON basis B
# express u in basis B
```

```python
v1 = np.array([-3/5,4/5,0])
v2 = np.array([4/5,3/5,0])
v3 = np.array([0,0,1])
B = [v1, v2, v3]
```

```python
dim_B = range(len(B))
isOrthonormal = True
for i in dim_B:
    for j in dim_B:
        if i == j and np.linalg.norm(B[i]) == 1:
            continue
        elif i == j and np.linalg.norm(B[i]) != 1.:
            isOrthonormal = False
        elif np.dot(B[i], B[j]) != 0:
            isOrthonormal = False
print(f"Basis B orthogonal:{isOrthonormal}")
```
