F(X) = BXa , where X belongs to S2 the space of all symmetrical 2x2 matrices

```python
import numpy as np
```

```python
B = np.array([
    [1, 0],
    [1, 1],
    [0, 1]
])
B
```

```python
a = np.array(
    [1,1])
a
```

```python
E1 = np.array([
    [1, 0],
    [0, 0]
])
E1
```

```python
E2 = np.array([
    [0, 0],
    [0, 1]
])
E2
```

```python
E3 = np.array([
    [0, 1],
    [1, 0]
])
E3
```

```python
E1_transformed = np.matmul(np.matmul(B, E1), a)
E1_transformed
```

```python
E2_transformed = np.matmul(np.matmul(B, E2), a)
E2_transformed
```

```python
E3_transformed = np.matmul(np.matmul(B, E3), a)
E3_transformed
```

```python
#matrix for F
F = np.linalg.matrix_transpose(np.array([E1_transformed, E2_transformed, E3_transformed]))
F
```

```python
np.matmul(F, [-1, -1, 1])
```
