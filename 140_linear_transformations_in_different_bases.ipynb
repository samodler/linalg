```python
import numpy as np
```

```python
u1 = np.array([-1, -1])
u2 = np.array([1, 2])
```

```python
transition_b_to_s = np.array([u1,u2]).transpose()
transition_b_to_s
```

```python
transition_s_to_b = np.linalg.inv(transition_b_to_s)
transition_s_to_b
```

```python
#Matrix of F in B
M = np.array([
    [2, 0],
    [0, -1]
])
M
```

```python
#F matrix of F in S
F = transition_b_to_s @ M @ transition_s_to_b
F
```
