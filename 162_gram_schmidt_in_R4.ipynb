```python
import numpy as np
```

```python
#basis vectors forming a subspace in R4
u1 = np.array([1,1,1,1])
u2 = np.array([4,2,4,2])
u3 = np.array([1,0,0,0])
```

```python
#Gram-Shmidt algorithm

v1 = u1 / np.linalg.norm(u1)
v2 = u2 - u2.dot(v1) * v1
v2 = v2 / np.linalg.norm(v2)
v3 = u3 - u3.dot(v2) * v2 - u3.dot(v1) * v1
v3 = v3 / np.linalg.norm(v3)
v1,v2,v3
```

```python
#Check v1-v3 form an ON basis
if np.dot(v1, v2) == np.dot(v1,v3) == np.dot(v2,v3) == 0:
    print('Vectors are pairwise orthogonal')
if np.linalg.norm(v1) == np.linalg.norm(v1) == np.linalg.norm(v1) == 1:
    print('All vectors are unitary')
```
