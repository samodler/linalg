```python
import numpy as np
```

```python
v = np.array([1, -2, 1])
v
```

```python
n = np.array([1, 1, 2])
n
```

```python
v_p = v - np.dot(n,v) / np.dot(n,n) * n
v_p
```

```python
T = np.identity(3) - np.outer(n, n) / np.dot(n,n)
T
```

```python
check = np.matmul(T, v)
np.allclose(check, v_p)
```

```python
det_T = np.linalg.det(T)
det_T
```

The determinant is zero as expected as we reduce the volume (dimension)
