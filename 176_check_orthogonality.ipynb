```python
import numpy as np
```

```python
sqrt = np.sqrt
A = np.array([
    [3/sqrt(11), -1/sqrt(6), -1/sqrt(66)],
    [1/sqrt(11), 2/sqrt(6), -4/sqrt(66)],
    [1/sqrt(11), 1/sqrt(6), 7/sqrt(66)],
])
A
```

```python
A_T = np.transpose(A)
A_T
```

```python
if np.allclose(A_T @ A, np.eye(3,3, dtype='float')) == True: 
    print('Matrix A is orthogonal')
```
