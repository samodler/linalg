```python
import numpy as np
```

```python
def check_eigenvector(A, v, tol=1e-8):
    # 1. Eigenvectors must be non-zero
    if np.allclose(v, 0, atol=tol):
        return False, None
    
    Av = A @ v
    
    # 2. Compute candidate eigenvalue, works for real and complex vectors
    lam = np.vdot(v, Av) / np.vdot(v, v)
    
    # 3. Check if A @ v is close to lambda * v
    is_eig = np.allclose(Av, lam * v, atol=tol)
    return is_eig, (lam if is_eig else None)
```

```python
A = np.array([
    [8, -9, 4],
    [3, -4, 3],
    [-3, 3, 1],
])
A
```

```python
v = np.array([1,2,3])
v
```

```python
is_eig, eigenvalue = check_eigenvector(A, v)
print(f"Is eigenvector: {is_eig}, Eigenvalue: {eigenvalue}")
# Output: Is eigenvector: True, Eigenvalue: 3.0
```
