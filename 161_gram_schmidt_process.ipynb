```python
import numpy as np
```

```python
f1 = np.array([2,4])
f2 = np.array([3,1])
```

```python
v1 = f1 / np.linalg.norm(f1)
print(f'v1={v1}, v1_norm = {np.linalg.norm(v1)}')
```

```python
v2 = f2 - np.dot(f2,v1) * v1
v2 = v2/np.linalg.norm(v2)
print(f'v1={v2}, v1_norm = {np.linalg.norm(v2)}')
```

```python
'Orthogonal' if np.dot(v1,v2) < 1e-10 else 'Not orthogonal'
```
