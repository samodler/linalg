Rotate vector v about x axis by angle alpha, following the righ hand rule.

```python
import numpy as np
```

```python
v = np.array([-2, 1, 2])
v
```

```python
norm = np.linalg.vector_norm(v)
norm
```

```python
alpha = np.pi / 6
alpha
```

```python
Rx = np.array([
    [1, 0, 0],
    [0, np.cos(alpha), -np.sin(alpha)],
    [0, np.cos(alpha), np.sin(alpha)]
])
Rx
```

```python
v_transformed = np.matmul(Rx, v)
v_transformed
```

```python
norm_after_transformation = np.linalg.vector_norm(v_transformed)
norm_after_transformation
```
