```python
import numpy as np
```

```python
norm = np.linalg.norm
```

```python
#Original basis vectors
u1 = np.array([-1,0,1,2])
u2 = np.array([0,1,0,1])
```

```python
#Check if original basis vectors form ON basis
'Vectors form an ON basis in W' \
    if (u1.dot(u2) == 0) & \
        (norm(u1) * norm(u2) == 1) \
        else 'Vectors do not form an ON basis in W'
```

```python
#Use Gram-Schmidt Process to orthogonolise the basis
v1 = u1 / norm(u1)
v2 = u2 - np.dot(u2,v1) * v1 
v2 = v2 / norm(v2)
v1, v2
```

```python
#Check if new basis vectors form ON basis
'Vectors form an ON basis in W' \
    if (v1.dot(v2) <= 1e-10) & \
        (norm(v1) * norm(v2) == 1) \
        else 'Vectors do not form an ON basis in W'
```

```python
#Find projection of a vector on the subspace
u = np.array([3, -4, 2, 5])
# w1 = proj_u_on_W
w1 = u.dot(v1) * v1 + u.dot(v2) * v2
# w2 = proj_u_Wperp
w2 = u - w1
w1, w2
```

```python
np.allclose(u, w1+w2,1e-10)
```

```python
#check w2 belongs to the orhogonal compliment Wperp
'v2 IS in Wperp' if w1.dot(w2) <= 1e-10 else 'v2 IS NOT in Wperp' 
```
