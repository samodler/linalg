```python
import numpy as np
```

```python
#vector in R^4
u = np.array([4, 3, 8, 1])
```

```python
#Projection on R3
#consider standard basis in R3
e1 = np.array([1,0,0,0])
e2 = np.array([0,1,0,0])
e3 = np.array([0,0,1,0])

proj_R3_u = u.dot(e1)/np.linalg.norm(e1)*e1 + \
    u.dot(e2)/np.linalg.norm(e2)*e2 + \
        u.dot(e3)/np.linalg.norm(e3)*e3
proj_R3_u
```
