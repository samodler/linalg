```python
import numpy as np
```

```python
#vector in R^3
u = np.array([4, 3, 2])
```

```python
#Projection on R2
#Vectors v1-v2 form a basis in R2
v1 = np.array([-1,2,3])
v2 = np.array([3,-1,2])

proj_R2_u = (u.dot(v1)/(np.linalg.norm(v1)**2))*v1 + \
    (u.dot(v2)/(np.linalg.norm(v2)**2))*v2
proj_R2_u
```

```python
orth_component =  u - proj_R2_u
orth_component
```
