```python
import numpy as np
```

```python
n = np.array([1, 1, 1])
n
```

```python
#rotation of the xy component of the normal vector
alpha = np.arctan(n[1] / n[0])
Rxy = np.array([
    [np.cos(alpha), np.sin(alpha), 0],
    [-np.sin(alpha), np.cos(alpha), 0],
    [0, 0, 1]
])
Rxy
```

```python
result_vector = np.matmul(Rxy, n)
formatted_vector = [f'{x:.3f}' for x in result_vector]
print(f'Vector: {formatted_vector}')
```

```python
#rotation of the rotated to xz plane vector to align with z axis
alpha = np.arctan(result_vector[0] / result_vector[2])
Rxz = np.array([
    [np.cos(alpha), 0, -np.sin(alpha)],
    [0, 1, 0],
    [np.sin(alpha), 0, np.cos(alpha)]
])
Rxz
```

```python
result_vector = np.matmul(Rxz, result_vector)
formatted_vector = [f'{x:.3f}' for x in result_vector]
print(f'Vector: {formatted_vector}')
```

```python
#Total rotation matrix
Q = np.matmul(Rxz, Rxy)
Q
```

```python
#reflect about xy plane
Sxy = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, -1]
])
Sxy
```

```python
#Total reflect matrix H = Q^T * Sxy * Q
H = np.matmul(np.matmul(np.transpose(Q), Sxy), Q)
H
```

```python
result_vector = np.matmul(H, n)
formatted_vector = [f'{x:.3f}' for x in result_vector]
print(f'Vector: {formatted_vector}')
```
