```python
import numpy as np
```

```python
v1 = np.array([-3/5, 4/5, 0])
v2 = np.array([4/5, 3/5, 0])
v3 = np.array([0, 0, 1])
```

```python
#check the norms
bool(np.linalg.norm(v1) == np.linalg.norm(v2) == np.linalg.norm(v3))
```

```python
#check the pair wise dot products are all 0
bool(np.dot(v1, v2) == np.dot(v1, v3) == np.dot(v2, v3))
```

v1, v2, v3 form an orthonormal basis B in R3.

```python
#vector u in the standard basis
u = np.array([1/7, -3/7, 5/7])
```

```python
#vector u in the B basis
u_B = np.array([np.dot(v1, u), np.dot(v2, u), np.dot(v3, u)])
u_B
```
