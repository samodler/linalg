```python
import numpy as np
```

```python
v1 = np.array([1, -2, 3, -4])
v2 = np.array([2, 1, -4, -3])
v3 = np.array([-3, 4, 1, -2])
v4 = np.array([4, 3, 2, 1])
new_basis = [v1, v2, v3, v4]
```

```python
isLinearlyIndependantSet = None

for i in range(len(new_basis)):
    for j in range(len(new_basis)):
        #print(f"checking dot(v{i}, v{j})")
        if j == i:
            #print(f"i == j")
            continue
        elif np.dot(new_basis[i], new_basis[j]) == 0:
            #print(f"i != j, dot(v{i}, v{j}) == 0")
            continue
        else:
            isLinearlyIndependantSet = False 
            #print(f"i != j, dot(v{i}, v{j}) != 0")
    isLinearlyIndependantSet = True
print(f"Set is linearly independant: {isLinearlyIndependantSet}")
```

```python
u = np.array([-1, 2, 3, 7])
u_coordinates_in_new_basis = []
for basis_vector in new_basis:
    u_coordinates_in_new_basis.append(np.dot(basis_vector, u)/(np.linalg.norm(basis_vector))**2)
u_coordinates_in_new_basis
```
